```python
import pandas as pd
```

```python
data = {
    'Name ': ['Alice ', 'Bob ', 'Charlie ', 'David '] ,
    'Age ': [24 , 19 , 22 , 25] ,
    'Score ': [88 , 92 , 85 , 95]}
df = pd . DataFrame ( data )
print(df)
```

```python
sorted_by_index=df.sort_index
print(sorted_by_index)
```

```python
first_two_rows=df.iloc[1:2]
print("First 2 rows: \n",first_two_rows)
```

```python
subset_loc = df.loc[df['Age '] > 20, ['Name ', 'Score ']]
print("Subset using loc:")
print(subset_loc)
```

```python
subset_brackets= df[['Name ', 'Age ']]
print("\n Subset using brackets:")
print(subset_brackets)
```

```python

name_column= df['Name ']
print("\n Name column:")
print(name_column)
```

```python
filter_single= df[df['Age ']>20]
print("\n Filter with single condition:")
print(filter_single)
```

```python
filtered_multiple= df[(df['Age ']>20) & (df['Score ']>90)]
print("\n Filter with multiple conditions:")
print(filtered_multiple)
```

```python
data={
    'Name': ['Alice','Bob','Charlie','David'],
    'Department':['HR','IT','Finance','HR'],
    'Score': [88,92,85,95]
}
df = pd . DataFrame ( data )
print(df)
```

```python
filtered_single=df[df['Department']=="HR"]
print(filtered_single)
```

```python
filtered_multiple=df[df['Department'].isin(['HR','IT'])]
print(filtered_multiple)
```

```python
data={
    'Category':['A','B','A','B','A'],
    'Value':[10,20,30,40,50]
}
df=pd.DataFrame(data)
grouped=df.groupby('Category')['Value'].mean()
print(grouped)
```

```python
df['Normalized']=df.groupby('Category')['Value'].transform(lambda x:x /x.sum())
print(df)
```

```python
filtered=df.groupby('Category').filter(lambda x:x['Value'].sum()>60)
print(filtered)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr'],
    'Sales': [200, 220, 250, 280]
}

df = pd.DataFrame(data)

# Line Plot
df.plot(x='Month', y='Sales', kind='line', marker='o', title='Monthly Sales')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(t):
    return np.exp(-t) * np.cos(2 * np.pi * t)

t1 = np.arange(0.0, 5.0, 0.1)
t2 = np.arange(0.0, 5.0, 0.02)

plt.figure()

plt.subplot(211)
plt.plot(t1, f(t1), 'bo', t2, f(t2), 'k')

plt.subplot(212)
plt.plot(t2, np.cos(2 * np.pi * t2), 'r--')

plt.show()
```
